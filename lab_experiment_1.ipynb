```python
df.dropna(inplace=True)
```

```python
df.isnull().sum()
```

```python
df.columns = df.columns.str.strip()

df['StudyHours'] = pd.to_numeric(df['StudyHours'], errors='coerce')
df['Attendance'] = pd.to_numeric(df['Attendance'], errors='coerce')
df['Marks'] = pd.to_numeric(df['Marks'], errors='coerce')

df['StudyHours'] = df['StudyHours'].fillna(df['StudyHours'].mean())
df['Attendance'] = df['Attendance'].fillna(df['Attendance'].mean())
df['Marks']=df['Marks'].fillna(df['Marks'].mean())
df = df.round()
print(df)
```

```python
print(f"Median of 'Marks' column: {df['Marks'].median()}")
df['Marks'] = df['Marks'].fillna(df['Marks'].median())
```

```python
print(df)
```

```python
df.columns = df.columns.str.strip()

df['Gender'] = df['Gender'].replace('', np.nan)

df['Gender'] = df['Gender'].fillna(df['Gender'].mode()[0])

print(df)
```

```python
df.duplicated().sum()
```

```python
df.columns = df.columns.str.strip()

df['Marks'] = pd.to_numeric(df['Marks'], errors='coerce')

df['Marks'] = df['Marks'].fillna(df['Marks'].mean())

print(df['Marks'].mean())
```

```python
plt.hist(df['Marks'],bins=10)
plt.xlabel('Marks')
plt.ylabel('Frequency')
plt.title('Marks Distribution')
plt.show()
```

```python
plt.boxplot(df['Marks'])
plt.ylabel('Marks')
plt.title("Box Plot of Marks")
```
