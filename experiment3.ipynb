```python
import pandas as pd

data = [
    ['green', 'hard', 'no',  'wrinkled', 'yes'],
    ['green', 'hard', 'yes', 'smooth',   'no'],
    ['brown', 'soft', 'no',  'wrinkled', 'no'],
    ['orange', 'hard', 'no', 'wrinkled', 'yes'],
    ['green', 'soft', 'yes', 'smooth',   'yes'],
    ['green', 'hard', 'yes', 'wrinkled', 'yes'],
    ['orange', 'hard', 'no', 'wrinkled', 'yes']
]


df = pd.DataFrame(
    data,
    columns=['color', 'hardness', 'texture', 'surface', 'target']
)

print("Dataset:")
print(df)

def find_s(data):

    X = data.iloc[:, :-1]
    y = data.iloc[:, -1]


    hypothesis = ['0'] * X.shape[1]

    for i in range(len(data)):

        if y.iloc[i] == 'yes':

            for j in range(X.shape[1]):


                if hypothesis[j] == '0':
                    hypothesis[j] = X.iloc[i, j]


                elif hypothesis[j] != X.iloc[i, j]:
                    hypothesis[j] = '?'

    return hypothesis


# Apply Find-S
final_hypothesis = find_s(df)

print("\nFinal Hypothesis:")
print(final_hypothesis)
```
