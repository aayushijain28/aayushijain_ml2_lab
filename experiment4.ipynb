```python
import pandas as pd

# Dataset
data = {
    "Sky": ["Sunny", "Sunny", "Rainy", "Sunny"],
    "Temperature": ["Warm", "Warm", "Cold", "Warm"],
    "Humidity": ["Normal", "High", "High", "Normal"],
    "Wind": ["Strong", "Strong", "Strong", "Weak"],
    "Enjoy": ["Yes", "Yes", "No", "Yes"]
}

df = pd.DataFrame(data)

# Separate features and target
X = df.iloc[:, :-1].values
y = df.iloc[:, -1].values

# Number of attributes
n = len(X[0])

# Initialize Specific boundary
S = ["∅"] * n

# Initialize General boundary
G = [["?"] * n]

print("Initial S:", S)
print("Initial G:", G)

for i, example in enumerate(X):

    print("\nExample", i + 1, ":", example, "->", y[i])

    # POSITIVE EXAMPLE
    if y[i] == "Yes":

        # Update S
        for j in range(n):

            if S[j] == "∅":
                S[j] = example[j]

            elif S[j] != example[j]:
                S[j] = "?"

        # Remove hypotheses from G that don't cover positive example
        G = [
            g for g in G
            if all(g[j] == "?" or g[j] == example[j] for j in range(n))
        ]

    # NEGATIVE EXAMPLE
    else:

        new_G = []

        for g in G:

            # If G covers the negative example, specialize it
            if all(g[j] == "?" or g[j] == example[j] for j in range(n)):

                for j in range(n):

                    # Create specialization based on S
                    if g[j] == "?" and S[j] != "?" and S[j] != "∅":

                        new_g = g.copy()
                        new_g[j] = S[j]

                        # Keep only hypotheses that are more general than S
                        if all(
                            new_g[k] == "?" or new_g[k] == S[k]
                            for k in range(n)
                        ):
                            new_G.append(new_g)

            else:
                new_G.append(g)

        G = new_G

    print("S:", S)
    print("G:", G)


print("\nFinal Specific Boundary (S):", S)
print("Final General Boundary (G):")

for g in G:
    print(g)
```

```python
# Candidate Elimination Algorithm - Negative Example Handling

import pandas as pd

data = {
    "Sky": ["Sunny", "Sunny", "Rainy"],
    "Temperature": ["Warm", "Warm", "Cold"],
    "Humidity": ["Normal", "High", "High"],
    "Wind": ["Strong", "Strong", "Strong"],
    "Enjoy": ["Yes", "Yes", "No"]
}

df = pd.DataFrame(data)

X = df.iloc[:, :-1].values
y = df.iloc[:, -1].values

n = len(X[0])

# Most Specific Hypothesis
S = ["∅"] * n

# Most General Hypothesis
G = [["?"] * n]

for i in range(len(X)):

    print("\nTraining Example:", X[i])
    print("Class:", y[i])

    # POSITIVE CASE
    if y[i] == "Yes":

        for j in range(n):

            if S[j] == "∅":
                S[j] = X[i][j]

            elif S[j] != X[i][j]:
                S[j] = "?"

        print("Positive Example")
        print("Updated S:", S)

    # NEGATIVE CASE
    else:

        print("Negative Example")

        new_G = []

        for g in G:

            # Check whether G covers negative example
            covers = True

            for j in range(n):

                if g[j] != "?" and g[j] != X[i][j]:
                    covers = False

            # If G covers negative example, specialize it
            if covers:

                for j in range(n):

                    # Specialize only where S has a specific value
                    if g[j] == "?" and S[j] != "?" and S[j] != "∅":

                        new_g = g.copy()
                        new_g[j] = S[j]

                        new_G.append(new_g)

            else:
                new_G.append(g)

        G = new_G

        print("S remains:", S)
        print("Updated G:", G)


print("\nFinal Specific Boundary:")
print(S)

print("\nFinal General Boundary:")
for g in G:
    print(g)
```
